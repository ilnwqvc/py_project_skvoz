```python
import pandas as pd
from pathlib import Path
from datetime import datetime

df = pd.read_csv(r"D:\progscode_2semestr\python_programming\data\normalized\variant_06\2026-03-23_22-16-41.csv")
df["ts"] = pd.to_datetime(df["ts"])

df["date"] = df["ts"].dt.date

ref = pd.read_csv(r"D:\progscode_2semestr\python_programming\reference\cities.csv")

print("before:", df.shape)
df = df.merge(ref, on="city_id", how="left")
print("after:", df.shape)

mart = df.groupby(["date","city_id","city_name"], as_index=False).agg(
    T_mean=("temperature_2m","mean"),
    P_sum=("precipitation","sum"),
    wind_max=("wind_speed_10m","max"),
    rainy_hours=("precipitation", lambda x:(x>0).sum())
)

print(mart.head())

now = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
path = Path(r"D:\progscode_2semestr\python_programming\data\mart\variant_06")
path.mkdir(parents=True, exist_ok=True)

mart.to_csv(path / f"mart_daily_{now}.csv", index=False)
```
